```python
import numpy as np
import cv2
import os

# ── Config ─────────────────────────────────────────────────────────
PATCH_DIR = "neqr_shares_eba\patch_256x256_set_A" # your patch output folder
ROWS, COLS = 256, 256

# ── Load shares ────────────────────────────────────────────────────
S1 = np.load(os.path.join(PATCH_DIR, 'S1.npy')).astype(np.uint16)
S2 = np.load(os.path.join(PATCH_DIR, 'S2.npy')).astype(np.uint16)
S3 = np.load(os.path.join(PATCH_DIR, 'S3.npy')).astype(np.uint16)

# ── Code 1 reverse logic ───────────────────────────────────────────
R1 = np.zeros((ROWS, COLS), np.uint16)
R2 = np.zeros((ROWS, COLS), np.uint16)

for b in range(4):
    R1 |= (((S1 >> (b+4)) & 1) ^ ((S2 >> b) & 1)) << b
    R1 |= ((S1 >> b) & 1) << (b+4)

for b in range(4):
    R2 |= (((S1 >> (b+4)) & 1) ^ ((S2 >> (b+4)) & 1) ^ ((S3 >> b) & 1)) << b
for b in range(4, 8):
    R2 |= ((S3 >> b) & 1) << b

# Extract correctly recovered MSB nibble for display
G1_disp = ((R1 & 0x0F) << 4).astype(np.uint8)
G2_disp = ((R2 & 0x0F) << 4).astype(np.uint8)

# ── Save ───────────────────────────────────────────────────────────
cv2.imwrite(os.path.join(PATCH_DIR, 'G1_recovered.png'), G1_disp)
cv2.imwrite(os.path.join(PATCH_DIR, 'G2_recovered.png'), G2_disp)
print("Recovered images saved.")
```
