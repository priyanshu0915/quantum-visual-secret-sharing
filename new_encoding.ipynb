```python
import matplotlib
matplotlib.use('Agg')
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyArrowPatch

fig, ax = plt.subplots(figsize=(14, 7))
ax.set_xlim(0, 14)
ax.set_ylim(0, 7)
ax.axis('off')
fig.patch.set_facecolor('white')

# ── colour scheme ────────────────────────────────────────────────
COL_SEC  = '#2196F3'   # blue  — secrets
COL_COV  = '#4CAF50'   # green — covers
COL_SHA  = '#9C27B0'   # purple — shares
COL_LSB  = '#FF9800'   # orange — LSB nibble
COL_MSB  = '#4CAF50'   # green  — MSB nibble
COL_XOR  = '#F44336'   # red    — XOR gate

def box(ax, x, y, w, h, color, label, fontsize=9, alpha=0.15):
    ax.add_patch(mpatches.FancyBboxPatch(
        (x, y), w, h,
        boxstyle="round,pad=0.05",
        facecolor=color, alpha=alpha,
        edgecolor=color, linewidth=2))
    ax.text(x + w/2, y + h/2, label,
            ha='center', va='center',
            fontsize=fontsize, fontweight='bold', color=color)

def arrow(ax, x1, y1, x2, y2, color='#555555'):
    ax.annotate('', xy=(x2, y2), xytext=(x1, y1),
                arrowprops=dict(arrowstyle='->', color=color,
                                lw=1.8))

def xor_gate(ax, x, y, color=COL_XOR):
    circle = plt.Circle((x, y), 0.22,
                         color=color, fill=False, linewidth=2)
    ax.add_patch(circle)
    ax.plot([x-0.22, x+0.22], [y, y], color=color, lw=2)
    ax.plot([x, x], [y-0.22, y+0.22], color=color, lw=2)

# ── Input column ─────────────────────────────────────────────────
# Secrets
box(ax, 0.3, 5.2, 1.4, 0.7, COL_SEC, '$G_1$\n(8-bit)')
box(ax, 0.3, 4.2, 1.4, 0.7, COL_SEC, '$G_2$\n(8-bit)')
# Covers
box(ax, 0.3, 2.8, 1.4, 0.7, COL_COV, '$C_1$\n(8-bit)')
box(ax, 0.3, 1.9, 1.4, 0.7, COL_COV, '$C_2$\n(8-bit)')
box(ax, 0.3, 1.0, 1.4, 0.7, COL_COV, '$C_3$\n(8-bit)')

ax.text(1.0, 6.2, 'NEQR Encoded\nInputs',
        ha='center', fontsize=8, color='#333333', style='italic')

# ── Nibble split labels ──────────────────────────────────────────
ax.text(2.3, 6.5, 'Nibble Split Logic (CNOT gates)',
        ha='center', fontsize=10, fontweight='bold', color='#333333')

# ── S1 construction ──────────────────────────────────────────────
# LSB nibble of S1 = G1[LSB] XOR G2[LSB]
arrow(ax, 1.7, 5.55, 3.1, 4.55, COL_LSB)
arrow(ax, 1.7, 4.55, 3.1, 4.55, COL_LSB)
xor_gate(ax, 3.3, 4.55, COL_LSB)
ax.text(3.3, 4.15, 'LSB\nnibble', ha='center', fontsize=7, color=COL_LSB)

# MSB nibble of S1 = C1[MSB]
arrow(ax, 1.7, 3.15, 3.1, 3.55, COL_MSB)
ax.text(3.3, 3.75, 'MSB\nnibble', ha='center', fontsize=7, color=COL_MSB)

# S1 box
box(ax, 3.8, 3.3, 1.4, 1.5, COL_SHA, '$S_1$\nLSB: $G_1 \\oplus G_2$\nMSB: $C_1$', fontsize=8)
arrow(ax, 3.55, 4.55, 3.8, 4.3, COL_LSB)
arrow(ax, 3.3, 3.55, 3.8, 3.8, COL_MSB)

# ── S2 construction ──────────────────────────────────────────────
arrow(ax, 1.7, 5.55, 6.1, 5.05, COL_LSB)   # G1 MSB
arrow(ax, 5.2, 4.3,  6.1, 4.55, COL_LSB)   # S1 MSB
xor_gate(ax, 6.3, 4.85, COL_LSB)
ax.text(6.3, 4.45, 'LSB\nnibble', ha='center', fontsize=7, color=COL_LSB)

arrow(ax, 1.7, 2.25, 6.1, 3.55, COL_MSB)   # C2 MSB
ax.text(6.3, 3.75, 'MSB\nnibble', ha='center', fontsize=7, color=COL_MSB)

box(ax, 6.8, 3.3, 1.4, 1.5, COL_SHA, '$S_2$\nLSB: $S_1 \\oplus G_1$\nMSB: $C_2$', fontsize=8)
arrow(ax, 6.55, 4.85, 6.8, 4.3, COL_LSB)
arrow(ax, 6.3,  3.55, 6.8, 3.8, COL_MSB)

# ── S3 construction ──────────────────────────────────────────────
arrow(ax, 1.7, 4.55, 9.1, 5.25, COL_LSB)   # G2 MSB
arrow(ax, 5.2, 4.3,  9.1, 4.95, COL_LSB)   # S1 MSB
arrow(ax, 8.2, 4.3,  9.1, 4.65, COL_LSB)   # S2 MSB
xor_gate(ax, 9.3, 4.95, COL_LSB)
ax.text(9.3, 4.45, 'LSB\nnibble', ha='center', fontsize=7, color=COL_LSB)

arrow(ax, 1.7, 1.35, 9.1, 3.55, COL_MSB)   # C3 MSB
ax.text(9.3, 3.75, 'MSB\nnibble', ha='center', fontsize=7, color=COL_MSB)

box(ax, 9.8, 3.3, 1.55, 1.5, COL_SHA,
    '$S_3$\nLSB: $S_1{\\oplus}S_2{\\oplus}G_2$\nMSB: $C_3$', fontsize=8)
arrow(ax, 9.55, 4.95, 9.8, 4.3,  COL_LSB)
arrow(ax, 9.3,  3.55, 9.8, 3.8,  COL_MSB)

# ── Auth tagging ─────────────────────────────────────────────────
box(ax, 11.7, 4.6, 1.9, 0.6, '#F44336', 'QPA\n(8 CNOTs/share)', fontsize=8)
box(ax, 11.7, 3.8, 1.9, 0.6, '#FF9800', 'EBA\n(H+2 CNOTs/share)', fontsize=8)
arrow(ax, 11.35, 4.05, 11.7, 4.9,  '#F44336')
arrow(ax, 11.35, 4.05, 11.7, 4.1,  '#FF9800')

ax.text(12.65, 3.45, 'Auth tags\n$\\mathcal{P}, \\mathcal{E}$\n(same shot)',
        ha='center', fontsize=8, color='#555555', style='italic')

# ── Legend ───────────────────────────────────────────────────────
legend_items = [
    mpatches.Patch(color=COL_SEC,  label='Secret image register'),
    mpatches.Patch(color=COL_COV,  label='Cover image register'),
    mpatches.Patch(color=COL_SHA,  label='Share register'),
    mpatches.Patch(color=COL_LSB,  label='LSB nibble (bits 0–3)'),
    mpatches.Patch(color=COL_MSB,  label='MSB nibble (bits 4–7)'),
]
ax.legend(handles=legend_items, loc='lower left',
          fontsize=8, framealpha=0.9, ncol=2)

ax.set_title('Share Generation via Nibble-Split CNOT Logic\n'
             r'$S_i[\mathrm{MSB}]$ carries cover image visuals  |  '
             r'$S_i[\mathrm{LSB}]$ carries XOR-mixed secret data',
             fontsize=11, fontweight='bold', color='#222222', pad=10)

plt.tight_layout()
plt.savefig('share_generation_schematic.png',
            dpi=300, bbox_inches='tight', facecolor='white')
plt.close()
print("Saved → share_generation_schematic.png")
```
